# Trajectory viewer

Start with `scripts/viewer.sh` on rdyu-server, tunnel from the laptop (`ssh -N -L 8888:localhost:8888 rdyu-server`), open the printed URL. `runs()` lists the local run directories with DCDs, `runs("rdyu-rog")` the ones on rdyu-rog (`node="rdyu-rog"` in `load` copies that DCD into `results/viewer/rdyu-rog/`). `color`: `"chain"`, `"z"` (pores) or `"type"`; hydrogens are hidden unless `hide_h=False`.

```python
import ovito_view as v

for node in ("local", "rdyu-rog"):
    print(node)
    for (system, stage), dcds in v.runs(node).items():
        print(f"  {system:32s} {stage:24s} {', '.join(dcds[:4])}{' ...' if len(dcds) > 4 else ''}")
```
